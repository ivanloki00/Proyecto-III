## ACTIVIDAD FINAL TEMA 3

#### BUSINESS GOAL (Objetivo de negocio)

- Identificar zonas con alta exposición a contaminación (hotspots) en Liverpool.

- Cuantificar el impacto en salud (mortalidad y enfermedades respiratorias) asociado a PM2.5 y PM10.

- Apoyar decisiones de planificación urbana (zonas de bajas emisiones, carriles bici, etc.).

- Detectar desigualdades ambientales en poblaciones vulnerables.

#### DATA (MODELLING) GOAL (Objetivo de datos / modelado)

- Desarrollar un modelo de Land Use Regression (LUR) para estimar niveles de contaminación.

- Analizar patrones temporales (hora del día, días de la semana, estaciones).

- Estimar la contaminación en toda la red de carreteras y agregarla a nivel LSOA.

- Relacionar exposición a contaminación con indicadores de salud.

#### MINABLE VIEW (Vista minable)

- Inputs:

    - Datos de sensores de calidad del aire (PM₂.₅, PM₁₀)

    - Red de carreteras (OSM)

    - Datos de tráfico (DfT)

    - Uso del suelo, densidad edificatoria, vegetación

    - Datos demográficos y de mortalidad (LSOA)

- Outputs:

    - Estimaciones de contaminación por segmento de carretera

    - Exposición media a nivel LSOA

    - Indicadores de impacto en salud

- Features:

    - Intensidad de tráfico

    - Clasificación viaria

    - Densidad de intersecciones

    - Uso del suelo

    - Cobertura verde

    - Densidad poblacional

#### CLARIFICACIÓN DE LA TAREA (Task)

- Problema de regresión (predicción de niveles de PM₂.₅ y PM₁₀).

- Con componente espacio-temporal.

#### MÉTRICA DEL MODELO (Model metric)

- $𝑅^2$

- RMSE (Root Mean Square Error)

- MAE (Mean Absolute Error)

#### MÉTRICA DE NEGOCIO (Business metric)

- Capacidad de identificar hotspots de contaminación

- Coherencia en la estimación de mortalidad atribuible

- Identificación de desigualdades socioeconómicas en exposición

#### HERRAMIENTAS (Tools)

- Python (Pandas, Scikit-learn)

- OSMNX (datos de red viaria)

- GIS / QGIS

#### TÉCNICAS (Techniques)

- Land Use Regression (LUR)

- Modelos de regresión (lineal, Random Forest, etc.)

- Análisis de series temporales

- Ingeniería de variables espaciales

- Validación cruzada espacial
