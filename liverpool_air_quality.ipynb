## 1. Contexto, valor e investigación 

### 1.1 Contexto del problema y relevancia científica
Este proyecto se sitúa en el escenario “hay datos externos que tienen valor para nosotros” ($out \rightarrow in$), ya que integra múltiples fuentes abiertas para generar conocimiento accionable sobre calidad del aire en Liverpool. Se utilizan mediciones de sensores roadside con frecuencia aproximada de 30 minutos, datos espaciales extraídos de OpenStreetMap mediante OSMNX, estadísticas de tráfico del DfT para calibración, información censal y de mortalidad a nivel de Lower Layer Super Output Area (LSOA), así como el Índice de Múltiple Privación.

El valor del proyecto no reside únicamente en predecir concentraciones de PM2.5 y PM10, sino en transformar datos ambientales heterogéneos en información útil para planificación urbana y salud pública. Este valor se construye a través de tres procesos fundamentales. En primer lugar, la curación y mejora de calidad de los datos, ya que los sensores de bajo coste pueden presentar sesgos, valores extremos o sensibilidad a condiciones meteorológicas como la humedad. En segundo lugar, la integración multiescala, combinando variabilidad temporal (hora, día, estación) con variabilidad espacial fina (segmentos de carretera) y agregada (barrios LSOA). Finalmente, la conversión de estimaciones ambientales en indicadores de impacto sanitario y desigualdad, permitiendo priorizar intervenciones en las zonas más vulnerables.

### 1.2 Resumen simple de investigación de dominio

La evidencia científica ha demostrado de forma consistente que la exposición crónica a material particulado fino (PM2.5 y PM10) está asociada a un mayor riesgo de mortalidad cardiovascular, enfermedad respiratoria y exacerbación del asma, especialmente en población vulnerable. Investigaciones recientes en epidemiología ambiental, como las lideradas por Francesca Dominici, han mostrado que incluso incrementos relativamente pequeños en la exposición a largo plazo a PM2.5 se asocian con aumentos estadísticamente significativos en mortalidad. Estos hallazgos refuerzan la necesidad de estimar la exposición con alta resolución espacial y de identificar posibles desigualdades ambientales.

En este contexto, los modelos de Land Use Regression (LUR) se han consolidado como una metodología estándar para explicar la variabilidad espacial de contaminantes atmosféricos a partir de características del entorno urbano, incluyendo intensidad de tráfico, clasificación viaria, densidad edificatoria, uso del suelo industrial y proporción de superficies vegetadas. Sin embargo, muchos estudios tradicionales se centran únicamente en medias anuales y en una única escala espacial, sin incorporar de manera explícita la dimensión temporal, pese a que la exposición real depende fuertemente del momento del día y de la estación. Además, los modelos existentes rara vez integran análisis de desigualdad socioeconómica, limitando su utilidad para evaluar impactos sanitarios y ambientales de manera integral.

Estos antecedentes justifican la necesidad de enfoques multiescala y temporalmente explícitos, que permitan no solo estimar concentraciones de PM en diferentes contextos urbanos, sino también vincularlas con la exposición poblacional y la carga sanitaria atribuible. Este marco de referencia proporciona la base científica que sustenta nuestro proyecto y asegura que las decisiones metodológicas se alineen con evidencia de alta calidad y mejores prácticas en epidemiología ambiental.

Los retos metodológicos son significativos y reflejan desafíos ampliamente reconocidos en estudios de contaminación urbana. Entre ellos se incluyen la variabilidad en la precisión de sensores de bajo coste, diferencias de resolución espacial entre datasets, riesgo de circularidad si se utilizan superficies modeladas como predictores, y la incertidumbre inherente a las funciones concentración-respuesta utilizadas para estimar efectos en salud. Reconocer estas limitaciones demuestra una comprensión crítica del dominio y proporciona la justificación científica para el diseño de estrategias multiescala y temporalmente explícitas en nuestro proyecto.

### 1.3 Definición de objetivos de negocio y datos

El objetivo general del proyecto es desarrollar un sistema multiescala de modelización espacio-temporal de la contaminación atmosférica en Liverpool que permita identificar periodos de máxima exposición, tramos de carretera de alto riesgo y barrios con impactos desproporcionados, así como estimar la carga sanitaria atribuible a la contaminación por partículas.

Desde el punto de vista aplicado, el objetivo de negocio es proporcionar evidencia que apoye decisiones de planificación urbana, protección de grupos vulnerables y reducción de desigualdades ambientales. Esto implica no solo identificar dónde hay más contaminación, sino también dónde coincide con mayor vulnerabilidad poblacional.

Desde el punto de vista de datos, los objetivos específicos incluyen la caracterización rigurosa de patrones temporales, el desarrollo de un modelo predictivo a nivel de segmento de carretera, la construcción de un modelo agregado a nivel LSOA y la integración de las estimaciones de contaminación con datos demográficos y de mortalidad para cuantificar impacto sanitario.

El proyecto se alinea con el ODS 3 (Salud y Bienestar), especialmente con el objetivo de reducir muertes asociadas a contaminación del aire, y con el ODS 11 (Ciudades y Comunidades Sostenibles), al contribuir a identificar y reducir desigualdades en exposición ambiental dentro del entorno urbano.

### 1.4 Criterios de éxito y métricas
El éxito del proyecto se evaluará en tres dimensiones complementarias. En primer lugar, el rendimiento predictivo de ambos modelos se medirá mediante métricas como R², RMSE y MAE, aplicando validación cruzada espacial para evitar sobreajuste derivado de autocorrelación geográfica. En segundo lugar, se considerará la capacidad del sistema para identificar de forma robusta segmentos de carretera y LSOAs situados en los percentiles superiores de exposición. Finalmente, se evaluará la capacidad del modelo LSOA para estimar de manera coherente la mortalidad atribuible y detectar gradientes socioeconómicos en la distribución de la contaminación.

## 3. Datos y Preparación

### 3.1 Informacion de la BBDD

El conjunto de datos principal utilizado en este proyecto corresponde a mediciones de material particulado registradas por sensores instalados en aproximadamente entre 40 y 70 localizaciones de Liverpool. Estos sensores registran concentraciones de contaminantes con una resolución temporal de aproximadamente 30 minutos, lo que permite analizar con gran detalle la variabilidad temporal de la contaminación atmosférica.
Cada observación del dataset representa una medición tomada por un sensor concreto en un momento específico. 

Concretamente, el conjunto de datos de sensores incluye variables de naturaleza espacial, temporal y ambiental.
Variables espaciales
Estas variables permiten identificar la localización de cada sensor:
-Sensor ID: identificador único de cada sensor.
-Latitud y longitud: coordenadas geográficas del punto de medición.
-Descripción de la ubicación (si está disponible): información textual sobre la localización del sensor.

Variables temporales
Cada medición incluye un timestamp que indica el momento exacto de registro.

Variables de contaminación
Corresponden a las concentraciones de material particulado, expresadas normalmente en microgramos por metro cúbico (µg/m³).
Representan la intensidad de contaminación en cada punto y momento.
-PM2.5: partículas finas con diámetro inferior a 2.5 µm
-PM10: partículas con diámetro inferior a 10 µm

Además del dataset de sensores, el proyecto incorpora múltiples fuentes de datos externas que se utilizarán como variables explicativas en los modelos de Land Use Regression (LUR). Entre estas fuentes destacan:
-Datos de OpenStreetMap (OSM) para usos del suelo, red viaria y huellas de edificios
-Datos de elevación procedentes de un Digital Terrain Model (DTM)
-Datos demográficos del Censo 2021 a nivel de Output Area y LSOA
-Estadísticas de mortalidad por causa específica a nivel de LSOA proporcionadas por la Office for National Statistics (ONS)


### 3.1 Inventario de fuentes y exploración preliminar
Listado de fuentes de datos y primeros pasos en la exploración de los mismos.

```python
import pandas as pd
import glob
import os

df_sensores = pd.read_csv("CoordsSensores.csv") 
df_sensores["sensor_id"] = df_sensores["sensor_id"].astype(str)

archivos = glob.glob("DatosCompletos/*.csv")
lista_df = []

for archivo in archivos:
    df = pd.read_csv(archivo)
    nombre = os.path.basename(archivo)
    partes = nombre.split("_")
    sensor_id = partes[2]
    df["sensor_id"] = sensor_id
    df["datetime"] = pd.to_datetime(df["Date & Time"])
    df = df.drop(columns=["Date & Time"])
    lista_df.append(df)

df_mediciones = pd.concat(lista_df, ignore_index=True)
df_mediciones["sensor_id"] = df_mediciones["sensor_id"].astype(str)

df_final = df_mediciones.merge(df_sensores, on="sensor_id", how="left")


columnas_interes = ["sensor_id", "datetime", "Temperature", "Humidity", "PM1.0", "PM2.5", "PM10", "lon", "lat"]
df_final = df_final[columnas_interes]

df_final = df_final.sort_values(["sensor_id", "datetime"]).reset_index(drop=True)

df_final.to_csv("dataset_limpio.csv", index=False)
print("Dataset final guardado en CSV:", df_final.shape)
```

```python
df = pd.read_csv("dataset_limpio.csv", dtype={"sensor_id": str}, low_memory=False)

cols_medicion = ["Temperature", "Humidity", "PM1.0", "PM2.5", "PM10"]

for col in cols_medicion:
    df[col] = pd.to_numeric(df[col], errors="coerce")

for col in cols_medicion:
    df[col] = df.groupby("sensor_id")[col].transform(lambda x: x.interpolate(method='linear'))

df.to_csv("dataset_limpio_interpolado.csv", index=False)
print("CSV limpio con nulos interpolados")
```

```python
import pandas as pd

df = pd.read_csv("dataset_limpio_interpolado.csv", dtype={"sensor_id": str}, low_memory=False)

df["Temperature"] = df["Temperature"].round(2)
df["Humidity"] = df["Humidity"].round(2)

df.to_csv("dataset_sensores.csv", index=False)
print("CSV de sensores completado")
```

## 4. Impacto, Riesgos y Ética

### 4.1 Análisis Coste-Beneficio (ROI) y vínculo con ODS/Sostenibilidad
El principal beneficio del proyecto es que puede ayudar a mejorar el conocimiento sobre la calidad del aire y su impacto en la salud pública, aunque estos beneficios suelen verse a largo plazo. Analizar cómo se distribuye la contaminación en Liverpool permite identificar qué zonas y momentos presentan mayores niveles de partículas contaminantes.

Los costes del proyecto son relativamente bajos y están relacionados principalmente con el uso de recursos computacionales y con el proceso de preparación de los datos. La limpieza y organización de los datos de sensores puede requerir bastante tiempo, especialmente cuando hay valores faltantes o errores.

Aun así, el proyecto puede aportar información útil para apoyar decisiones relacionadas con la planificación urbana o con la salud pública. Además, el proyecto está relacionado con el ODS 3 (Salud y Bienestar) y el ODS 11 (Ciudades y Comunidades Sostenibles), ya que contribuye a mejorar la comprensión de los problemas de contaminación en entornos urbanos.

### 4.2 Matriz de mitigación de riesgos
Durante el desarrollo del proyecto pueden aparecer diferentes riesgos. Uno de los principales es que los sensores no capturen toda la variación de la contaminación en la ciudad, ya que solo están situados en ciertos puntos.

Para reducir este problema se pueden usar modelos LUR para estimar valores en áreas donde no hay sensores.

Otro posible riesgo está relacionado con la calidad de los datos. Los sensores pueden registrar errores o valores incompletos, por lo que es necesario revisar y limpiar los datos antes de realizar el análisis.

### 4.3 Consideraciones éticas y sesgos del modelo
Desde el punto de vista ético, es importante asegurarse de que el modelo no genere resultados que puedan interpretarse de forma discriminatoria. Por ejemplo, es necesario analizar si algunas zonas aparecen sistemáticamente como más contaminadas y cómo se interpretan esos resultados.

También es posible que exista sesgo debido a la ubicación de los sensores. Si algunos barrios tienen menos sensores o menos datos disponibles, las estimaciones pueden ser menos precisas.

Por ello, los resultados deben interpretarse con cuidado y teniendo en cuenta las limitaciones de los datos utilizados.
