## 1. Contexto, Valor e Investigación (Domain & Value)

### 1.1 Identificación en el "Bazar de los datos"
En esta sección se debe clasificar el flujo de valor del proyecto. Dado que utilizamos datos externos (sensores, OSM, Censo) para generar valor interno y público en Liverpool, este proyecto se sitúa principalmente en el escenario **"That data out there is valuable for me ($out \rightarrow in$)"**. El modelo de monetización o valor se centra en **Predictive Action as a Service**, al predecir la calidad del aire para intervenciones de salud.

### 1.2 Resumen simple de investigación de dominio
Breve resumen de la investigación sobre la calidad del aire en Liverpool y su contexto.

### 1.3 Definición de objetivos de negocio y datos
Es crucial no confundir metas organizacionales con técnicas.

*   **Objetivo de Negocio:** Reducir la exposición de la población vulnerable y la mortalidad atribuible a la contaminación en Liverpool.
*   **Objetivo de Datos:** Desarrollar modelos de regresión (LUR) que estimen con precisión niveles de PM2.5 y PM10 a nivel de calle y barrio.

### 1.4 Criterios de éxito y métricas
El éxito no es solo "que el modelo funcione", debe ser cuantificable.

*   **Éxito de Negocio:** Identificación del 100% de los "hotspots" en zonas escolares.
*   **Éxito de Datos:** Lograr un error relativo (RSE) inferior al 20% o un $R^2$ específico en la validación cruzada.

## 4. Impacto, Riesgos y Ética

### 4.1 Análisis Coste-Beneficio (ROI) y vínculo con ODS/Sostenibilidad
El beneficio del proyecto es la mejora de la salud pública (intangible a largo plazo). Los costes incluyen recursos computacionales y el "traumático" proceso de data wrangling (limpieza de sensores), que suele subestimarse. El beneficio neto debe ser positivo al optimizar las intervenciones urbanas.

### 4.2 Matriz de mitigación de riesgos
Identificamos riesgos internos y externos. Por ejemplo:

*   **Riesgo:** Los sensores roadside no capturan toda la variación espacial.
*   **Mitigación:** Usar datos de tráfico de la DfT y modelos LUR para interpolar en áreas sin cobertura.

### 4.3 Consideraciones éticas y sesgos del modelo
Debemos asegurar que el modelo no sea intrínsecamente discriminatorio. Es vital analizar si las recomendaciones de salud afectan desproporcionadamente a ciertos barrios o si existe un sesgo en la ubicación original de los sensores que ignore zonas desfavorecidas (IMD).
