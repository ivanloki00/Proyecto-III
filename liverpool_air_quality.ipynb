## 1. Contexto, valor e investigación 

### 1.1 Identificación en el "Bazar de los datos"
En esta sección se debe clasificar el flujo de valor del proyecto. Dado que utilizamos datos externos (sensores, OSM, Censo) para generar valor interno y público en Liverpool, este proyecto se sitúa principalmente en el escenario **"Hay datos fuera que tienen valor para mi ($out \rightarrow in$)"**. El modelo de monetización o valor se centra en **Predictive Action as a Service**, al predecir la calidad del aire para intervenciones de salud.

**Valor del dato:** el valor de nuestros activos de datos no es estático, sino que crece mediante los siguientes procesos definidos en la metodología:
- Curación y Precisión: Los datos brutos de los sensores de Liverpool presentan desafíos de calibración y completitud; el proceso de control de calidad les añade valor.
- Conexión: El valor aumenta exponencialmente al conectar la calidad del aire con el entorno construido (OSM) y la vulnerabilidad de la población (Censo).
- Conversión en Insights y Decisiones: La meta final es transformar estas mediciones en un modelo de regresión de uso del suelo que soporte decisiones sobre planificación urbana y salud pública.

### 1.2 Resumen simple de investigación de dominio

Dado que se busca entender el modelo de regresión (que características geográficas y relaciones modelan las concentraciones de contaminantes) y obtener cierta funcionalidad de este, en esta sección se explica la base necesaria para abordar el analisis de la calidad del aire.
La eficacia del modelo depende de la obtención de predictores extraídos de **OpenStreetMap** (estadísticas de tráfico):
- Tráfico Rodado (highway=): Es el principal determinante de la variabilidad del material particulado. Se utilizan categorías como motorway o primary como aproximadores.
-Densidad Edificada (building=): Los edificios actúan como fuentes de emisión (calefacción) y como obstáculos que atrapan contaminantes en "cañones urbanos".
- Uso de Suelo Industrial: Crucial para capturar el impacto de zonas portuarias y áreas industriales específicas del norte de Liverpool.
El uso de **redes de sensores de bajo coste** (LCS) en una ciudad costera como Liverpool presenta retos críticos:
- Humedad Relativa (RH): A niveles superiores al 75%, las partículas absorben agua y los sensores ópticos informan concentraciones mucho más altas de las reales.
- Estrategia: Es imperativo aplicar modelos de corrección no lineales (como Random Forest) que integren variables meteorológicas para reducir el error de los datos crudos.
Por último hay que vincular la exposición con desenlaces clínicos mediante **Funciones de Respuesta a la Concentración**

### 1.3 Definición de objetivos de negocio y datos

*   **Objetivo de Negocio:** Reducir la exposición ciudadana, principalmete para proteger a los grupos vulnerables. mediante la identificación de las zonas criticas con grandes grupos sociales.
*   **Objetivo de Datos:** Desarrollar un modelo de regresión de uso del suelo que estime con precisión niveles de PM2.5 y PM10 a nivel de calle y barrio

Nuestro análisis no solo busca precisión estadística, sino que está alineado con la agenda global de sostenibilidad, por ello se muestra a continuación los SGD trakers alienados con el proyecto:
- ODS 3: Salud y Bienestar: **TRARGET 3.9** (*Reducir el número de muertes y enfermedades producidas por productos químicos peligrosos y la contaminación del aire*)

Para cumplir con este target, el proyecto no solo mapea contaminantes, sino que cuantifica el riesgo biológico real para la población de Liverpool.
- ODS 11: Sostenibilidad urbana: **TRARGET 11.6**  (*Detectar las zonas donde mayor exposición a la contaminación hay*)

Para cumplir con este target, el proyecto detectará directamente las zonas de la ciudad de Liverpool donde más cantidad de contaminación hay,  para alejar a la gente mas vulnerable de esas zonas.

### 1.4 Criterios de éxito y métricas
El éxito no es solo "que el modelo funcione", debe ser cuantificable.

*   **Éxito de Negocio:** Identificación del 100% de los "hotspots" en zonas escolares.

## 3. Datos y Preparación

### 3.1 Inventario de fuentes y exploración preliminar
Listado de fuentes de datos y primeros pasos en la exploración de los mismos.

```python
import pandas as pd
import glob
import os

df_sensores = pd.read_csv("CoordsSensores.csv") 
df_sensores["sensor_id"] = df_sensores["sensor_id"].astype(str)

archivos = glob.glob("DatosCompletos/*.csv")
lista_df = []

for archivo in archivos:
    df = pd.read_csv(archivo)
    nombre = os.path.basename(archivo)
    partes = nombre.split("_")
    sensor_id = partes[2]
    df["sensor_id"] = sensor_id
    df["datetime"] = pd.to_datetime(df["Date & Time"])
    df = df.drop(columns=["Date & Time"])
    lista_df.append(df)

df_mediciones = pd.concat(lista_df, ignore_index=True)
df_mediciones["sensor_id"] = df_mediciones["sensor_id"].astype(str)

df_final = df_mediciones.merge(df_sensores, on="sensor_id", how="left")


columnas_interes = ["sensor_id", "datetime", "Temperature", "Humidity", "PM1.0", "PM2.5", "PM10", "lon", "lat"]
df_final = df_final[columnas_interes]

df_final = df_final.sort_values(["sensor_id", "datetime"]).reset_index(drop=True)

df_final.to_csv("dataset_sensores_junto.csv", index=False)
print("Dataset guardado")
```

```python
import pandas as pd

df = pd.read_csv("dataset_sensores_junto.csv", dtype={"sensor_id": str}, low_memory=False)

cols_medicion = ["Temperature", "Humidity", "PM1.0", "PM2.5", "PM10"]

for col in cols_medicion:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Temperature"] = df["Temperature"].round(2)
df["Humidity"] = df["Humidity"].round(2)

for col in cols_medicion:
    df[col] = df.groupby("sensor_id")[col].transform(lambda x: x.interpolate(method='linear'))

df.to_csv("dataset_sensores.csv", index=False)
print("CSV limpio con nulos interpolados")
```
